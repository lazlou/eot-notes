```python
# Make a folder to save downloaded files
!mkdir -p EOT-2020/parquet/subsets/
```

## 5. Download a Single Test File

```python
# Download part-00046
!aws s3 cp --no-sign-request s3://eotarchive/eot-index/table/eot-main/crawl=EOT-2020/part-00046-dda73194-fd75-4dcb-b361-4d099d882262-c000.gz.parquet EOT-2020/parquet/subsets/
```

## 6. Load and Inspect the Parquet File

```python
import pandas as pd

# Load the downloaded Parquet file
df = pd.read_parquet("EOT-2020/parquet/subsets/part-00046-dda73194-fd75-4dcb-b361-4d099d882262-c000.gz.parquet")

# Show column names and types
print(df.dtypes)

# Preview the first few rows
df.head()
```
